In this assignment, we analyze whether the existing cost model used by the company is still accurate. Restaurant owners have reported that their actual weekly operating costs are higher than the model suggests. To verify this claim, we use hypothesis testing, which helps us compare expected values with real-world data and decide whether the difference is significant or not.

## Understanding the Problem

The company uses a formula to calculate weekly cost based on production units. According to this formula, the cost depends on a fixed value and a variable cost per unit. However, restaurant owners believe that the actual cost is higher than what the formula predicts. Our task is to test this claim using statistical methods. We will compare the expected cost with the observed sample data and determine whether the difference is meaningful or just due to random variation.

## Step 1: Defining Hypothesis

In hypothesis testing, we start by defining two assumptions. The first is the null hypothesis, which assumes that the existing cost model is correct and there is no increase in cost. The second is the alternative hypothesis, which assumes that the actual cost is higher than the model. These two assumptions help us structure the problem and decide what we are trying to prove using the data.

# Null Hypothesis (H0): The cost model is correct (no increase in cost)
# Alternative Hypothesis (H1): Actual cost is higher than the model

## Step 2: Calculating Expected Cost

In this step, we calculate the expected weekly cost using the given formula. The formula includes a fixed cost and a variable cost per unit produced. By substituting the number of units into the formula, we find the expected cost according to the model. This value will later be compared with the actual observed cost from the sample data to check whether there is a difference.

```python
# Given value
X = 600

# Cost formula
expected_cost = 1000 + 5 * X
expected_cost
```

## Step 3: Calculating Test Statistic

To compare the expected cost with the actual sample cost, we calculate the Z value. This value tells us how far the sample mean is from the expected mean in terms of standard error. A large difference indicates that the observed result is not likely due to random variation. This step is important because it gives a numerical value that helps us make a decision about the hypothesis.

```python
import numpy as np

# Given values
sample_mean = 3050
population_mean = expected_cost
sigma = 5 * 25
n = 25

# Standard Error
SE = sigma / np.sqrt(n)

# Z value
Z = (sample_mean - population_mean) / SE
Z
```

## Step 4: Decision Making

In this step, we compare the calculated Z value with the critical value at a 5 percent significance level. The critical value defines the acceptable range where the difference is considered normal. If the calculated value lies outside this range, it indicates that the difference is too large to ignore. This helps us decide whether to accept or reject the null hypothesis.

```python
# Critical value for 5% significance level
critical_value = 1.96

# Decision check
abs(Z) > critical_value
```

## Conclusion

The calculated Z value is significantly different from the critical value, which means the observed sample cost does not match the expected cost from the model. Therefore, we reject the null hypothesis. This indicates that the existing cost model is not accurate anymore. The data supports the claim made by restaurant owners that the actual weekly operating costs are higher than what the model suggests.
