In this assignment, we estimate the average durability of print-heads used in a company. Since testing each print-head destroys it, only a small sample is available. To solve this problem, we use confidence intervals, which help us estimate a range where the true average value is likely to fall. This method allows us to make reliable decisions even when we do not have complete data for the entire population.

## Understanding the Problem

The company wants to know how long their print-heads last, but they cannot test all units because testing destroys them. Instead, they test a small sample of 15 print-heads. Our task is to use this sample data to estimate the average durability of all print-heads. We will calculate a confidence interval, which provides a range of values that likely contains the true average. This helps in making decisions without needing full data.

## Step 1: Loading the Data

In this step, we input the sample data into the notebook. Each value in the dataset represents the durability of a single print-head measured in millions of characters. Loading the data correctly is important because all further calculations depend on this dataset. Once the data is loaded, we can begin analyzing it and calculating the required statistical measures.

```python
import numpy as np

data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

data
```

## Step 2: Calculating the Mean

The mean represents the average durability of the print-heads in the sample. It is calculated by adding all values and dividing by the number of observations. This gives us a central value that represents the overall performance of the sample. The mean is important because it is used as the base value in confidence interval calculations.

```python
mean = np.mean(data)
mean
```

## Step 3: Calculating Standard Deviation

Standard deviation measures how much the data values vary from the average. A small value indicates that the data points are close to the mean, while a large value indicates more variation. This measure is important because it helps us understand how spread out the data is, and it plays a key role in calculating the confidence interval.

```python
std = np.std(data, ddof=1)
std
```

## Step 4: Confidence Interval using t-distribution

Since the sample size is small and the population standard deviation is unknown, we use the t-distribution. This method adjusts for small sample sizes and gives a more accurate range. The confidence interval tells us the range in which the true average durability is likely to fall with a certain level of confidence, such as 99 percent.

```python
import scipy.stats as stats

n = len(data)

# t value for 99% confidence
t_value = stats.t.ppf(0.995, df=n-1)

margin = t_value * (std / np.sqrt(n))

lower = mean - margin
upper = mean + margin

(lower, upper)
```

## Step 5: Confidence Interval using Z-distribution

In this case, we assume that the population standard deviation is known. When this value is available, we use the Z-distribution instead of the t-distribution. This simplifies the calculation and still provides a reliable range. This step helps us understand how results differ when more information about the population is available.

```python
sigma = 0.2
z_value = 2.576  # for 99% confidence

margin = z_value * (sigma / np.sqrt(n))

lower = mean - margin
upper = mean + margin

(lower, upper)
```

## Final Interpretation

The confidence interval provides a range of values within which the true average durability of print-heads is likely to lie. Instead of giving a single value, this method gives a more realistic estimate by considering uncertainty in the data. A higher confidence level means a wider range but more reliability. This approach is useful in real-world situations where complete data is not available.
