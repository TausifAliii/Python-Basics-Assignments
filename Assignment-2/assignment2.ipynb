Q1: Create a list of 5 integers and perform append, insert, remove, pop, and reverse operations.

In this question, we created a list of integers and applied different list operations.
The append() method adds an element at the end of the list, insert() adds an element at a specific position,
remove() deletes a specific value, pop() removes the last element, and reverse() changes the order of elements.
These operations help in modifying list data dynamically.

```python
def list_operations():
    numbers = [10, 20, 30, 40, 50]

    numbers.append(60)
    numbers.insert(2, 25)
    numbers.remove(40)
    numbers.pop()
    numbers.reverse()

    print("Final List:", numbers)

list_operations()
```

This code uses different list methods to modify the list such as append, insert, remove, pop, and reverse.

Q2: Create a tuple and try to change one value.

A tuple is a data structure that stores multiple values, but it is immutable.
This means once a tuple is created, its elements cannot be changed.
When we try to modify an element, Python raises a TypeError.
This ensures data safety when values should not be altered.

```python
def tuple_example():
    names = ("Alice", "Bob", "Charlie")

    try:
        names[1] = "David"
    except TypeError:
        print("Tuples cannot be modified")

tuple_example()
```

Tuples are immutable, which means their values cannot be changed after creation.

Q3: Create two sets and perform union, intersection, and difference.

Sets are unordered collections that do not allow duplicate values.
In this example, union combines elements from both sets,
intersection finds common elements, and difference finds elements present in one set but not in the other.
These operations are useful when comparing datasets.

```python
def set_operations():
    set1 = {1, 2, 3, 4}
    set2 = {3, 4, 5, 6}

    print("Union:", set1 | set2)
    print("Intersection:", set1 & set2)
    print("Difference:", set1 - set2)

set_operations()
```

Set operations like union, intersection, and difference are used to compare elements between two sets.

Q6: Calculate grade based on marks.

Explanation:
In this problem, multiple conditions are checked using if-elif-else statements.
Based on the marks, different grades are assigned.
This type of conditional logic is useful when there are multiple possible outcomes.

```python
def calculate_grade(marks):
    if marks >= 90:
        print("Grade: A")
    elif marks >= 75:
        print("Grade: B")
    elif marks >= 50:
        print("Grade: C")
    else:
        print("Fail")

calculate_grade(85)
```

Q7: Check if number is positive, even, odd, or negative.

Explanation:
This code checks multiple conditions.
First, it determines whether the number is positive, negative, or zero.
If the number is positive, it further checks whether it is even or odd using the modulus operator (%).
This demonstrates nested conditional statements.

```python
def number_check(num):
    if num > 0:
        if num % 2 == 0:
            print("Positive Even")
        else:
            print("Positive Odd")
    elif num < 0:
        print("Negative Number")
    else:
        print("Zero")

number_check(10)
```

The code checks if the number is positive, negative, or zero and also checks if it is even or odd.

Q8: Create scalar, 1D array, and 2D array.

Explanation:
NumPy is used for numerical operations and array handling.
A scalar represents a single value, a 1D array is like a list, and a 2D array is like a matrix.
The reshape() function is used to convert a 1D array into a 2D structure.

```python
def numpy_arrays():
    scalar = np.array(5)
    arr_1d = np.arange(1, 6)
    arr_2d = np.arange(10, 70, 10).reshape(2, 3)

    print("Scalar:", scalar)
    print("1D Array:", arr_1d)
    print("2D Array:\n", arr_2d)

numpy_arrays()
```

NumPy is used to create arrays. A scalar is a single value, 1D array is a list of values, and 2D array is a matrix.

Q10: Convert NumPy array to Pandas DataFrame.

Explanation:

Q11: Perform operations on a DataFrame.

Explanation:
This program demonstrates basic data handling using Pandas.
A DataFrame is created with sample data.
The info() function shows dataset details, and describe() provides statistical summary.
Specific columns are selected, unnecessary columns are removed, missing values are filled using the mean,
and duplicate rows are removed.
These steps are part of data cleaning and preprocessing.

```python
import pandas as pd

# Function to perform DataFrame operations
def pandas_operations():
    data = {
        "Name": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
        "Age": [23, 25, 22, 24, 26, 28, 27, 29, 30, 31],
        "City": ["NY", "LA", "SF", "TX", "NY", "LA", "SF", "TX", "NY", "LA"],
        "Salary": [50000, 60000, None, 55000, 65000, None, 70000, 72000, 68000, 75000]
    }

    # Create DataFrame
    df = pd.DataFrame(data)

    # Display basic information
    print("Info:")
    df.info()

    # Display statistical summary
    print("\nSummary:")
    print(df.describe())

    # Select specific columns
    print("\nSelected Columns (Name and City):")
    print(df[["Name", "City"]])

    # Drop 'City' column
    df = df.drop("City", axis=1)

    # Fill missing values in Salary column (correct way, no warning)
    mean_salary = df["Salary"].mean()
    df["Salary"] = df["Salary"].fillna(mean_salary)

    # Remove duplicate rows
    df = df.drop_duplicates()

    # Display final DataFrame
    print("\nFinal DataFrame:")
    print(df)

# Call function
pandas_operations()
```


This code creates a DataFrame, shows information and summary, selects columns, removes a column, fills missing values, and removes duplicates.
