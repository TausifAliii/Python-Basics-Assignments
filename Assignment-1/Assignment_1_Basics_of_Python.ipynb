```python
#2. Variables and Data Types
#A variable is used to store data in Python.
#Each variable has a data type, which tells what kind of value it is storing.
#Example code:

age = 22            # integer
salary = 45000.50   # float
name = "Tausif"     # string
is_student = True   # boolean

print(age)
print(salary)
print(name)
print(is_student)
```

```python
#3. Input and Output
#Input is taken from the user using the input() function and output is displayed using print().
#sample code:

name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Name:", name)
print("Age:", age)
```

```python
#4. Operators in Python
#Arithmetic operators are used to perform mathematical operations.
#sample code :

a = 10
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
#5. Conditional Statements
#Conditional statements are used to make decisions in a program.
#code:

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is Even")
else:
    print("The number is Odd")
```

```python
#6. Loops

#Loops are used to repeat a block of code multiple times.
#They help in reducing code repetition and make programs shorter.

# For Loop
#The for loop is used when the number of iterations is known in advance.
#In this example, the loop prints numbers from 1 to 5.

 # While Loop
#The while loop is used when the number of iterations is not fixed.
#It runs until the given condition becomes false.
#code:

# For Loop
for i in range(1, 6):
    print(i)

# While Loop
i = 1
while i <= 5:
    print(i)
    i += 1

```

```python
#7. Functions
# Functions are used to divide a program into smaller reusable parts.
# They make the code easier to understand and manage.
#code:

def add(a, b):
    return a + b

result = add(5, 3)
print("Sum:", result)
```

```python
#8. Lists
#A list is used to store multiple values in a single variable.
#It is mutable, which means its values can be changed.
#code:

numbers = [10, 20, 30, 40]
print(numbers)
print("First element:", numbers[0])
```

```python
#9. Tuples
#A tuple is similar to a list but it is immutable.
#Once created, its values cannot be modified
#code:

fruits = ("apple", "banana", "mango")
print(fruits)
```

```python
#10. Dictionary
#A dictionary stores data in key-value pairs.
#It is used when data needs to be stored in a structured form.
#sample code:
student = {
"name": "Tausif",
"age": 22,
"course": "Data Science"
}

print("Student Name:", student["name"])
```

```python
#11. Conditional Program (Even/Odd)
#This program checks whether a number is even or odd.
#If the number is divisible by 2, it is even; otherwise, it is odd.
#sample code:
# Conditional Program (Even/Odd)

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Number is Even")
else:
    print("Number is Odd")
```
