## Handling Missing Values

In this dataset, missing values are represented using the symbol '?'. These values cannot be used directly, so we replace them with NaN and remove the rows containing missing values. This ensures that the dataset becomes clean and suitable for machine learning models. Proper handling of missing data is important because incomplete data can affect model performance.

```python
df = df.replace("?", np.nan)
df = df.dropna()
```

## Cleaning Column Names

Column names may contain spaces or special characters, which can cause errors during coding. In this step, we clean the column names by replacing spaces and special characters with underscores. This makes them easier to use and improves code readability.

```python
df.columns = df.columns.str.strip().str.replace(" ", "_").str.replace("-", "_")
```

```python
# Label Encoding
le = LabelEncoder()
if 'occupation' in df.columns:
    df['occupation'] = le.fit_transform(df['occupation'])

# One-Hot Encoding
df = pd.get_dummies(df, drop_first=True)
```

## Feature Scaling

Different numerical features may have different ranges, which can affect model performance. Feature scaling ensures that all numerical values are on a similar scale. Standard Scaling transforms data to have mean zero and standard deviation one. Min-Max Scaling converts values between 0 and 1. The choice of scaling depends on the model and data distribution.

```python
num_cols = df.select_dtypes(include=np.number).columns

# Standard Scaling
scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])

# Min-Max Scaling (demonstration)
minmax = MinMaxScaler()
df[num_cols] = minmax.fit_transform(df[num_cols])
```

## Feature Engineering

Feature engineering involves creating new features that help the model better understand the data. In this step, we create a new feature called age_hours by combining age and working hours. We also create capital_total, which represents the net financial gain by subtracting capital loss from capital gain. These new features provide additional insights and improve prediction accuracy.

```python
df['age_hours'] = df['age'] * df['hours_per_week']
df['capital_total'] = df['capital_gain'] - df['capital_loss']

# Target variable
df['income_binary'] = df.filter(like='income_').iloc[:, 0]
```

## Log Transformation

Some numerical features, such as capital gain, may have very large values, which creates a skewed distribution. Log transformation reduces this skewness by compressing large values. This helps the model learn patterns more effectively and reduces the impact of extreme values.

```python
df['capital_gain_log'] = np.log1p(df['capital_gain'])
```

## Outlier Detection using Isolation Forest

Outliers are unusual data points that differ significantly from the rest of the dataset. These points can negatively affect model performance. Isolation Forest is used to detect such outliers by isolating abnormal data points. Removing or identifying outliers helps improve model accuracy and reliability.

```python
iso = IsolationForest(contamination=0.01, random_state=42)
df['outliers'] = iso.fit_predict(df[num_cols])
```
